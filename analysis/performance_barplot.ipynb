```python
import pandas as pd
import matplotlib.pyplot as plt

# Define the data
data = {
    'Accuracy': [0.8989, 0.9464, 0.8828],
    'F1-score': [0.9124, 0.9538, 0.9270],
    'Precision': [0.9083, 1.0000, 0.9474],
    'Recall': [0.9167, 0.9118, 0.9076],
    'Specificity': [0.8750, 1.0000, 0.7692]
}
index = [
    'ADNI-Train \n(Multi-visits)',
    'ADNI-Test \n(Multi-visits)',
    'ADNI-Test \n(Single-visit)'
]

# Create DataFrame
df = pd.DataFrame(data, index=index)

# Plotting
ax = df.T.plot(kind='bar', figsize=(9, 3), rot=0)

plt.title('Classification Performance: CN (centiloid<10) vs. AD')

# Keep legend position unchanged
plt.legend(title='Dataset', bbox_to_anchor=(1.05, 1), loc='upper left')

# Y-axis range
plt.ylim(0, 1)

# Grid
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator, FormatStrFormatter

# Define the data
data = {
    'Accuracy': [0.8227],
    'F1-score': [0.7619],
    'Precision': [0.6452],
    'Recall': [0.9302],
    'Specificity': [0.7755]
}
index = ['NACC_single']

# Create DataFrame
df = pd.DataFrame(data, index=index)

fig, ax = plt.subplots(figsize=(3.6, 3), dpi=200)

# Plotting
df.T.plot(
    kind='bar',
    ax=ax,
    rot=0,
    legend=False,
    fontsize=8,
    width=0.75,
    color="#4C72B0",
    edgecolor="white",
    linewidth=0.6
)

plt.title('Classification Performance: CN (centiloid<10) vs. AD', fontsize=9)

# Y-axis
plt.ylim(0, 1.0)
ax.yaxis.set_major_locator(MultipleLocator(0.2))
ax.yaxis.set_major_formatter(FormatStrFormatter('%.1f'))

# Grid
plt.grid(axis='y', linestyle='--', alpha=0.7, linewidth=0.6)
ax.set_axisbelow(True)

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Optional: keep left/bottom cleaner
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

plt.tight_layout()
plt.show()
```
